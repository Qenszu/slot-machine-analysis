```python
PATH = '../data/casino_sim.db'
conn = sqlite3.connect(PATH)
```

```python
df = pd.read_sql_query("""
        SELECT player_id, SUM(bet) as bets,  SUM(win) as wins, SUM(bet) - SUM(win) as profit
        FROM spins
        GROUP BY player_id
                       """, conn)

df
```
